    # In Colab, asyncio.run() can cause RuntimeError if an event loop is already running.
    # Instead, we can directly await the main function if not running as a standalone script.
    # For standalone execution, asyncio.run(main()) is correct.
    # In Colab's interactive environment, top-level await is often supported or
    # the following can be used if main is meant to be run in the background:
    # loop = asyncio.get_event_loop()
    # loop.run_until_complete(main())
    # However, for simplicity and common Colab usage, let's assume it can be directly awaited
    # or that the user will run it in an appropriate context. The original error is about
    # calling run() when one is already there, so we remove the run().
    # If the user intends to run this as a script, they would need to manage the loop.
    # For interactive Colab execution, we will remove asyncio.run()

```python
import aiohttp
import asyncio


def do_requests(session):
    return session.get('https://example.com')


async def demo():
    async with aiohttp.ClientSession() as session:
        tasks = []
        for _ in range(0, 10):
            tasks.append(do_requests(session))

        results = await asyncio.gather(*tasks)
        for r in results:
            print('example.com =>', r.status)
```

```python
asyncio.run(demo())
```

```python
await demo()
```
