```python
with open('simple.txt', 'w') as f:
    f.write('使用「推薦碼」推薦好友加入 MM 訂閱會員，雙方皆可以獲得 [M幣](https://www.macromicro.me/user/mcoins)！')
```

```python
from google import genai
from google.genai import types
client = genai.Client(api_key='AIzaXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXX00AA')
```

```python
list(client.file_search_stores.list())
```

```python
file_search_store = client.file_search_stores.create(config={'display_name': 'TEST'})
print(f"Created store: {file_search_store.name}")
```

```python
client.file_search_stores.upload_to_file_search_store(
    file='simple.txt',
    file_search_store_name=file_search_store.name,
)
```

```python
list(client.file_search_stores.list())
```

```python
response = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents="兌換碼",
    config=types.GenerateContentConfig(
        tools=[
            types.Tool(
                file_search=types.FileSearch(
                    file_search_store_names=[file_search_store.name]
                )
            )
        ]
    )
)
```

```python
from pprint import pprint as pp
pp(response.model_dump())
```

```python
client.file_search_stores.delete(name=file_search_store.name, config=types.DeleteFileSearchStoreConfig(force=True))
```
