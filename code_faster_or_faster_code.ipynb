```python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    else:
        return fibonacci_recursive(n-1) + fibonacci_recursive(n-2)

def fibonacci_iterative(n):
    if n <= 1:
        return n
    else:
        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        return b
```

```python
def run(fibonacci, number_demo=40):
    print(f"Calculating the {number_demo}th Fibonacci number using recursion...")

    import time
    start_time = time.time()
    result = fibonacci(number_demo)
    end_time = time.time()

    elapsed_time = end_time - start_time

    print(f"The {number_demo}th Fibonacci number is: {result}")
    print(f"Time taken for {number_demo}th item: {elapsed_time:.5f} seconds")
    print()

run(fibonacci_recursive)
run(fibonacci_iterative)
```
