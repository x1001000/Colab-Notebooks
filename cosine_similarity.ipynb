> Y : {array-like, sparse matrix} of shape (n_samples_Y, n_features), default=None
Input data. If `None`, the output will be the pairwise similarities between all samples in `X`.

```python
from google import genai
from google.genai import types
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

client = genai.Client()
```

```python
texts = [
    '條紋',
    '襯衫',
    '條紋襯衫',
    '無條紋的襯衫',
]
```

```python
result = client.models.embed_content(
    model="gemini-embedding-001",
    contents=texts,
    config=types.EmbedContentConfig(task_type="SEMANTIC_SIMILARITY")
)

embeddings = [e.values for e in result.embeddings]
similarity_matrix = cosine_similarity(embeddings)

print('gemini-embedding-001 嵌入模型\n\n的輸出向量為例\n\n四個詞彙的兩兩相似度由高至低\n')

similarity_results = []
for i, text1 in enumerate(texts):
    for j in range(i + 1, len(texts)):
        text2 = texts[j]
        similarity = similarity_matrix[i, j]
        similarity_results.append((similarity, text1, text2))
similarity_results.sort(key=lambda x: x[0], reverse=True)

similarity_df = pd.DataFrame(similarity_results, columns=['Similarity', 'Text 1', 'Text 2'])
display(similarity_df)
```
