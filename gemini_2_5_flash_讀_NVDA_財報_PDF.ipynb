```python
prompt = '''
Extract values from https://nvidianews.nvidia.com/_gallery/download_pdf/6837703d3d63320fddb3a9ee/
Output values in JSON format.

keys:
company_name
currency
period_end_date
quarter
units
account_payables
cash_and_cash_equivalents
cash_and_short_term_investments
common_stock
cost_of_revenue
depreciation_and_amortization
effect_of_forex_change_on_cash
eps
eps_diluted
free_cash_flow
goodwill_and_intangible_assets
gross_profit
income_before_tax
income_tax_expense
interest_expense
interest_income
inventory
long_term_debt
net_cash_provided_by_operating_activities
net_cash_used_for_investing_activities
net_cash_used_provided_by_financing_activities
net_change_in_cash
net_debt
net_income
net_receivables
operating_expenses
operating_income
preferred_stock
property_plant_equipment_net
research_and_development_expenses
retained_earnings
revenue
selling_general_and_administrative_expenses
short_term_debt
short_term_investments
total_assets
total_current_assets
total_non_current_assets
total_liabilities
total_current_liabilities
total_non_current_liabilities
total_debt
total_equity
'''
```

```python
from google import genai
from google.genai.types import Tool, GenerateContentConfig
from google.colab import userdata
client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
    config=GenerateContentConfig(tools=[{"url_context": {}}])
)
```

```python
import re
json_string = re.search(r'\{.*\}', response.text, re.DOTALL).group(0)

import json
key_values = json.loads(json_string)

import pandas as pd
pd.Series(key_values)
```
